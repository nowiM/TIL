```python
import matplotlib.pyplot as plt

# test_data에서 컬럼 이름 'Date'를 'ds'로 변경 (Prophet 예측 결과와 일치시킴)
test_data = test_data.rename(columns={'Date': 'ds', 'Close': 'y'})

# 날짜 필터 조건 설정
start_date = '2025-01-02'
end_date = '2025-06-27'

# forecast와 test_data에서 해당 기간만 필터링
forecast_range = forecast[(forecast['ds'] >= start_date) & (forecast['ds'] <= end_date)][['ds', 'yhat']]
test_range = test_data[(test_data['ds'] >= start_date) & (test_data['ds'] <= end_date)][['ds', 'y']]

# 날짜 기준 병합
merged = pd.merge(forecast_range, test_range, on='ds', how='inner')

# 시각화
plt.figure(figsize=(12, 6))
plt.plot(merged['ds'], merged['y'], label='Actual', marker='o')
plt.plot(merged['ds'], merged['yhat'], label='Predicted', linestyle='--')
plt.title('Prophet Forecast vs Actual (2025-01-02 ~ 2025-06-27)')
plt.xlabel('Date')
plt.ylabel('Value')
plt.legend()
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from prophet import Prophet
import matplotlib.pyplot as plt

# 1. 데이터 로딩 (이미 train_data 존재한다고 가정)
# train_data = pd.read_csv('your_data.csv')  # 필요 시 사용

# 2. Prophet용 컬럼만 추출
df = train_data[['ds', 'y']].copy()

# 3. 로그 변환 (주가 분산 안정화 및 추세 완화)
df['y'] = np.log(df['y'])

# 4. Prophet 모델 정의 및 학습
model = Prophet(
    daily_seasonality=True,
    weekly_seasonality=True,
    yearly_seasonality=True,
    changepoint_prior_scale=0.1  # 추세 변화 감지 민감도 (0.01 ~ 0.5)
)
model.fit(df)

# 5. 미래 날짜 생성 (예: 30일 예측)
future = model.make_future_dataframe(periods=30)

# 6. 예측
forecast = model.predict(future)

# 7. 로그 복원
forecast['yhat_original'] = np.exp(forecast['yhat'])
forecast['yhat_lower_original'] = np.exp(forecast['yhat_lower'])
forecast['yhat_upper_original'] = np.exp(forecast['yhat_upper'])

# 8. 시각화 (복원된 값 기준)
plt.figure(figsize=(14, 6))
plt.plot(df['ds'], np.exp(df['y']), label='Actual')
plt.plot(forecast['ds'], forecast['yhat_original'], label='Forecast')
plt.fill_between(forecast['ds'], forecast['yhat_lower_original'], forecast['yhat_upper_original'], 
                 color='lightblue', alpha=0.4, label='Confidence Interval')
plt.title('Prophet Forecast (Original Scale)')
plt.xlabel('Date')
plt.ylabel('Close Price')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
