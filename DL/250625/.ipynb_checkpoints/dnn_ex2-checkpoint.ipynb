```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def numerical_derivative(f, x): # 수치 미분 함수
    delta_x = 1e-4
    grad = np.zeros_like(x)
    it = np.nditer(x, flags =['multi_index'], op_flags = ['readwrite'])
    
    while not it.finished:
        idx = it.multi_index
        tmp_val = x[idx]
        x[idx] = float(tmp_val) + delta_x
        fx1 = f(x) #
        x[idx] = tmp_val - delta_x
        fx2 = f(x)
        grad[idx] = (fx1 - fx2) / (2 * delta_x)
        x[idx] = tmp_val
        it.iternext()

    return grad
```

```python
import numpy as np

class LogicGate:
    def __init__(self, gate_name, xdata, tdata):
        self.name = gate_name
        self.xdata = xdata.reshape(4, 2)
        self.tdata = tdata.reshape(4, 1)

        # 은닉층
        self.W2 = np.random.rand(2, 6)
        self.b2 = np.random.rand(6) # 바이어스

        # 출력층
        self.W3 = np.random.rand(6, 1)
        self.b3 = np.random.rand(1) #바이어스
        
        self.learning_rate = 1e-2

    def loss_val(self):
        delta = 1e-7
        # 은닉층 손실값 계산
        z2 = np.dot(self.xdata, self.W2) + self.b2
        a2 = sigmoid(z2)
        #출력층 손실값 계산
        z3 = np.dot(a2, self.W3) + self.b3
        y = a3 = sigmoid(z3)
        
        return -np.sum(self.tdata * np.log(y + delta) + (1 - self.tdata) * np.log((1 - y) + delta))
        
    def train(self):
        f = lambda x : self.loss_val()
        print("Initial loss value = ", self.loss_val())
        for step in range(15001):
            # 은닉층 미분계산
            self.W2 -= self.learning_rate * numerical_derivative(f, self.W2)
            self.b2 -= self.learning_rate * numerical_derivative(f, self.b2)
            # 출력층 미분계산
            self.W3 -= self.learning_rate * numerical_derivative(f, self.W3)
            self.b3 -= self.learning_rate * numerical_derivative(f, self.b3)
            if (step % 1000 == 0):
                print("step = ", step, "loss value = ", self.loss_val())
                
            

    def predict(self, input_data):
        self.xdata = input_data
        z2 = np.dot(input_data, self.W2) + self.b2
        a2 = sigmoid(z2)
        z3 = np.dot(a2, self.W3) + self.b3
        y = a3 = sigmoid(z3)

        if y > 0.5:
            result = 1
        else:
            result = 0

        return y, result

    def
```

# 다층 퍼셉트론

```python
import numpy as np

class LogicGate:
    def __init__(self, gate_name, xdata, tdata):
        self.name = gate_name
        self.xdata = xdata.reshape(4, 2)
        self.tdata = tdata.reshape(4, 1)

        # 은닉층
        self.hidden_1 = np.random.rand(2, 4)
        self.hidden_1_bise = np.random.rand(4) # 바이어스

        self.hidden_2 = np.random.rand(4, 2)
        self.hidden_2_bise = np.random.rand(2) # 바이어스

        # 출력층
        self.out = np.random.rand(2, 1) # W3
        self.out_bise = np.random.rand(1) #바이어스 b3
        
        self.learning_rate = 1e-2

    def loss_val(self):
        delta = 1e-7
        z2 = np.dot(self.xdata, self.hidden_1) + self.hidden_1_bise
        a2 = sigmoid(z2)
        
        z3 = np.dot(a2, self.hidden_2) + self.hidden_2_bise
        a3 = sigmoid(z3)
        
        z4 = np.dot(a3, self.out) + self.out_bise
        y = a4 = sigmoid(z4)
        
        return -np.sum(self.tdata * np.log(y + delta) + (1 - self.tdata) * np.log((1 - y) + delta))
        
    def train(self):
        f = lambda x : self.loss_val()
        print("Initial loss value = ", self.loss_val())
        for step in range(50001):
            # 은닉층 미분계산
            self.hidden_1 -= self.learning_rate * numerical_derivative(f, self.hidden_1)
            self.hidden_1_bise -= self.learning_rate * numerical_derivative(f, self.hidden_1_bise)

            self.hidden_2 -= self.learning_rate * numerical_derivative(f, self.hidden_2)
            self.hidden_2_bise -= self.learning_rate * numerical_derivative(f, self.hidden_2_bise)
            
            # 출력층 미분계산
            self.out -= self.learning_rate * numerical_derivative(f, self.out)
            self.out_bise -= self.learning_rate * numerical_derivative(f, self.out_bise)
            if (step % 1000 == 0):
                print("step = ", step, "loss value = ", self.loss_val())
                
            

    def predict(self, input_data):
        self.xdata = input_data
        z2 = np.dot(self.xdata, self.hidden_1) + self.hidden_1_bise
        a2 = sigmoid(z2)
        
        z3 = np.dot(a2, self.hidden_2) + self.hidden_2_bise
        a3 = sigmoid(z3)
        
        z4 = np.dot(a3, self.out) + self.out_bise
        y = a4 = sigmoid(z4)

        if y > 0.5:
            result = 1
        else:
            result = 0

        return y, result
```

```python
xdata = np.array([ [0, 0], [0, 1], [1, 0], [1, 1] ])
tdata = np.array([0, 1, 1, 0])

xor_obj = LogicGate('XOR', xdata, tdata)
xor_obj.train()
```

```python
test_data = np.array([ [0, 0], [0, 1], [1, 0], [1, 1] ])

for data in test_data:
    sigmoid_val, logical_val = xor_obj.predict(data)
    print(data, ' = ', logical_val)
```
