```python
import os
import time
from ultralytics import YOLO
```

```python
trained_model = YOLO("runs/classify/my_custom_train4/weights/best.pt")
```

```python
folders = [
    "datasets/polygon/val/images/heptagon",
    "datasets/polygon/val/images/hexagon",
    "datasets/polygon/val/images/octagon"
]
```

```python
valid_exts = (".jpg")
output_txt_path = "inference_results.txt"
total_inference_time = 0.0
total_images = 0
```

```python
with open(output_txt_path, "w", encoding="utf-8") as f:
    for folder_path in folders:
        f.write(f"\n=== Inference in folder: {folder_path} ===\n")
        print(f"\n=== Inference in folder: {folder_path} ===")

        image_files = [img for img in os.listdir(folder_path) if img.lower().endswith(valid_exts)]

        for img_file in image_files:
            img_path = os.path.join(folder_path, img_file)
            start_time = time.time()

            results = trained_model.predict(img_path)

            end_time = time.time()
            inference_time = end_time - start_time
            total_inference_time += inference_time
            total_images += 1

            top1_index = results[0].probs.top1
            top1_conf = float(results[0].probs.top1conf)

            print(f"Image: {img_path}")
            print(f" - Predicted Class Index: {top1_index}")
            print(f" - Confidence: {top1_conf:.4f}")
            print(f" - Inference Time: {inference_time:.4f} s")

            f.write(f"\nImage: {img_path}\n")
            f.write(f" - Predicted Class Index: {top1_index}\n")
            f.write(f" - Confidence: {top1_conf:.4f}\n")
            f.write(f" - Inference Time: {inference_time:.4f} s\n")

    f.write("\n==================== SUMMARY ====================\n")
    print("\n==================== SUMMARY ====================")

    if total_images > 0:
        average_time = total_inference_time / total_images
        summary_str = (
            f"Total Images: {total_images}\n"
            f"Total Inference Time: {total_inference_time:.4f} s\n"
            f"Average Inference Time per Image: {average_time:.4f} s\n"
        )
    else:
        summary_str = "No images were found for inference.\n"

    f.write(summary_str)
    print(summary_str)

print(f"Inference results saved to: {output_txt_path}")
```
